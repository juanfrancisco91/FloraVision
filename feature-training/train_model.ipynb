# 🌸 FloraVision — Entrenamiento de Modelo IA (`feature/model-training`)
Este notebook se encarga de realizar el entrenamiento de Transfer Learning y Fine-Tuning para la clasificación inteligente de flores y estado fisiológico en el proyecto **FloraVision**.

### 📌 Objetivos:
1. **Cargar modelo preentrenado** (`MobileNetV2` / `ResNet50` sobre ImageNet).
2. **Aplicar Data Augmentation** (giros, rotación, zoom, contraste, traslación).
3. **Entrenar con el Dataset** en dos fases (Transfer Learning + Fine-Tuning).
4. **Exportar el modelo final entrenado** a un archivo (`modelo_flores.h5` y `modelo_flores.keras`).

```python
import os
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, optimizers, callbacks

print(f"TensorFlow Version: {tf.__version__}")
print(f"GPU Disponibles: {len(tf.config.list_physical_devices('GPU'))}")
```

## 1. Configuración de Rutas y Clases de Flores

```python
CLASES_FLORES = ["rosa", "girasol", "margarita", "clavel", "lirio", "orquidea", "tulipan"]
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
EPOCHS = 10

BASE_DIR = Path.cwd().parent if Path.cwd().name == "feature-training" else Path.cwd()
DATASET_DIR = BASE_DIR / "dataset"
MODEL_OUTPUT_H5 = BASE_DIR / "modelo_flores.h5"
MODEL_OUTPUT_KERAS = BASE_DIR / "modelo_flores.keras"

print(f"Directorio Base: {BASE_DIR}")
print(f"Ruta del Dataset: {DATASET_DIR}")
```

## 2. Preparación y Carga del Dataset con Data Augmentation

```python
# Importar la función auxiliar del módulo train_model
sys.path.append(str(BASE_DIR / "feature-training"))
from train_model import preparar_dataset_muestra, obtener_capa_data_augmentation

# Inicializar dataset si no existe
if not DATASET_DIR.exists():
    preparar_dataset_muestra(DATASET_DIR)

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical"
)

# Optimización de rendimiento I/O
train_ds = train_ds.cache().shuffle(1000).prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=tf.data.AUTOTUNE)
```

## 3. Demostración de Data Augmentation

```python
data_aug = obtener_capa_data_augmentation()
plt.figure(figsize=(10, 6))
for images, labels in train_ds.take(1):
    first_image = images[0]
    for i in range(6):
        ax = plt.subplot(2, 3, i + 1)
        augmented_image = data_aug(tf.expand_dims(first_image, 0))
        plt.imshow(augmented_image[0].numpy().astype("uint8"))
        plt.title(f"Augmentation #{i+1}")
        plt.axis("off")
plt.suptitle("Visualización de Data Augmentation Aplicado a la Imagen Base")
plt.tight_layout()
plt.show()
```

## 4. Construcción del Modelo con MobileNetV2 / ResNet50

```python
from train_model import construir_modelo_flores

model, base_model = construir_modelo_flores(model_type="mobilenet", num_classes=len(CLASES_FLORES))
model.summary()
```

## 5. Entrenamiento FASE 1: Transfer Learning (Backbone Congelado)

```python
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

cb_list = [
    callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True),
    callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, verbose=1),
    callbacks.ModelCheckpoint(filepath=str(MODEL_OUTPUT_H5), monitor="val_accuracy", save_best_only=True)
]

history_p1 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=cb_list
)
```

## 6. Entrenamiento FASE 2: Fine-Tuning (Ajuste Fino de Capas Superiores)

```python
base_model.trainable = True
fine_tune_at = len(base_model.layers) // 2
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

history_p2 = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=max(2, EPOCHS // 2),
    callbacks=cb_list
)
```

## 7. Exportación del Modelo Entrenado (`modelo_flores.h5`)

```python
model.save(MODEL_OUTPUT_H5)
print(f"✅ Modelo exportado exitosamente a H5: {MODEL_OUTPUT_H5}")

try:
    model.save(MODEL_OUTPUT_KERAS)
    print(f"✅ Modelo exportado exitosamente a Keras: {MODEL_OUTPUT_KERAS}")
except Exception as e:
    print(f"ℹ️ Nota sobre archivo .keras: {e}")
```

## 8. Verificación de Inferencia en Imagen de Prueba

```python
img_path = BASE_DIR / "feature-capture" / "imagenes_prueba" / "rosa_roja.jpg"
if img_path.exists():
    img = keras.utils.load_img(img_path, target_size=IMG_SIZE)
    img_array = keras.utils.img_to_array(img)
    img_batch = tf.expand_dims(img_array, 0)
    
    predictions = model.predict(img_batch)
    predicted_class = CLASES_FLORES[np.argmax(predictions[0])]
    confidence = np.max(predictions[0]) * 100
    
    plt.imshow(img)
    plt.title(f"Predicción IA: {predicted_class.capitalize()} ({confidence:.1f}% de confianza)")
    plt.axis("off")
    plt.show()
else:
    print("Imagen de prueba no encontrada para demostración final.")
```
